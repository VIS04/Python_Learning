# OOPS

```python
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
    def area (self):
        return(self.length*self.breadth)
    def perimeter (self):
        return(2*(self.length+self.breadth))
```

```python
rec1 = Rectangle(4,2)
```

```python
print(rec1.length)
print(rec1.breadth)
```

```python
rec1.area()
```

```python
rec1.perimeter()
```
