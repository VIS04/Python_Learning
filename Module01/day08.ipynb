# Error Handling

```python
try:
    num = int(input('enter a number'))
    ans = (100/num)
except ZeroDivisionError:
    print('division by zero is not possible')
except ValueError:
    print('only enter a number')
else:
     print(ans)
finally:
    print('the function ran completely, thank you')
```

```python
def divide_by_100():
    try:
        num = int(input('enter a number'))
        ans = (100/num)
    except ZeroDivisionError:
        print('division by zero is not possible')
    except ValueError:
        print('only enter a number')
    else:
        print(ans)
    finally:
        print('the function ran completely, thank you')
```

```python
divide_by_100()
```

```python
def divide_by_100():
    while True:
        try:
            num = int(input('enter a number'))
            ans = (100/num)
        except ZeroDivisionError:
            print('division by zero is not possible')
        except ValueError:
            print('only enter a number')
        else:
            print(ans)
            break # Exit
        finally:
            print('the function ran completely, thank you')
```

```python
divide_by_100()
```
