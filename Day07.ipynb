# Function

```python
def oddoreven(num): #parameter
    """This function is used to identify 
    odd and even number"""
    if num%2 ==0:
        print('even') #just show/display value
    else:
        return('odd') #have memory allocation, can be called to another function.
```

```python
oddoreven(4) #argument
```

```python
num1 = int(input('enter a number'))
oddoreven(num1)
```

```python
def minmax(numbers):
    return min(numbers), max(numbers)
```

```python
minmax([1,2,3,4,5,6])
```

```python
a,b = minmax([1,2,3,4,5,6])
print (a)
print (b)
```

```python
def add (a,b):
    return a+b
```

```python
add (1,2)
```

```python
def add (*a): # *args (positional arguments), input takes an tupple
    return sum(a)
```

```python
add(1,2,3,4)
```

```python
def detail(**info):#**kwargs, input take as dictionary
    print(type(info))
    for key, val in info.items():
        print(f'{key}:{val}')
```

```python
detail(name='visakh', status = 'student')
```

```python
def greetings(name,wish):
    print(f'{wish} {name}')
```

```python
greetings('visakh', 'hi')
```

```python
greetings('hi', 'visakh')
```

# lambda function

```python
h = [1,2,3,4,5] #iterable
def square(x): #function
    return(x**2)
```

```python
list(map(square,h)) #map (function, iterable)
```

```python
tuple(map(square,h))
```

```python
map(square,h) #map object created, not listed, need container to store
```

```python
list(map(square,h)) # container -> map(function, iterable)
```

```python
list(map(lambda x:x**2,h)) #list comprehension
```

```python
list(filter(lambda x:x%2==0,h))
```
