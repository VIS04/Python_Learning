```python
#18 LCM of 2 numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

num1 = a
num2 = b

while num2 != 0:
    temp = num2
    num2 = num1 % num2
    num1 = temp

gcd = num1
lcm = (a * b) // gcd

print("LCM =", lcm)
```

```python
#20 Reverse a number
num = int(input("Enter a number: "))

rev = 0
n = abs(num)

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n //= 10

if num < 0:
    rev = -rev

print("Reversed number =", rev)
```

```python
#27 Amstrong Number
num = int(input("Enter a number: "))

power = len(str(num))
temp = num
total = 0

while temp > 0:
    digit = temp % 10
    total += digit ** power
    temp //= 10

if total == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
#28 Palindrome
num = int(input("Enter a number: "))

temp = num
rev = 0

while temp > 0:
    digit = temp % 10
    rev = rev * 10 + digit
    temp //= 10

if rev == num:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#29 Strong Number
num = int(input("Enter a number: "))

temp = num
total = 0

while temp > 0:
    digit = temp % 10
    fact = 1
    for i in range(1, digit + 1):
        fact *= i
    total += fact
    temp //= 10

if total == num:
    print("Strong Number")
else:
    print("Not Strong Number")
```

```python
#32 Prime numbers in range
start = int(input("Enter start: "))
end = int(input("Enter end: "))

print("Prime numbers:")

for num in range(start, end + 1):
    if num > 1:
        prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                prime = False
                break
        if prime:
            print(num)
```

```python
#37 Hollow Square pattern
n = int(input("Enter size: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
#38 Pascals traingle
n = int(input("Enter rows: "))

for i in range(n):
    num = 1
    for j in range(n - i):
        print(" ", end="")
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print()
```

```python
#40 ATM
balance = 1000

while True:
    print("\n--- ATM Menu ---")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        print("Balance =", balance)

    elif choice == 2:
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print("Deposit successful.")

    elif choice == 3:
        amount = float(input("Enter withdrawal amount: "))
        if amount <= balance:
            balance -= amount
            print("Withdrawal successful.")
        else:
            print("Insufficient balance.")

    elif choice == 4:
        print("Thank you for using the ATM.")
        break

    else:
        print("Invalid choice.")
```
