```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,20,15,30,25]
```

```python
plt.title("demo graph")
plt.xlabel("marks")
plt.ylabel("data")
plt.plot(x,y)
plt.show()
```

```python
plt.bar(x,y)
```

```python
plt.scatter(x,y)
```

```python
plt.hist(y)
```

```python
plt.pie(x)
plt.show()
```
