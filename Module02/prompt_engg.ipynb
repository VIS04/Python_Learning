# Prompt Engineering

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()
```

```python
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
MODEL = "gpt-5-mini"
```

```python
#simple chatbot
while True:
    user_input = input("\nYou: ")

    if user_input.lower()=="exit":
        break

    response = client.responses.create(
        model = MODEL,
        input =user_input

    )
    print("\n GPT: ", response.output_text)
```

```python
response = client.responses.create(
        model = MODEL,
        input ="what is AI"
)
print("\n GPT: ", response.output_text)
```

```python
#zero shot (no example in prompt)

response = client.responses.create(
        model = MODEL,
        input ="classify the sentiment: 'the delivery was late for 2 days' "
)
print("\n GPT: ", response.output_text)
```

```python
#one/few shot

response = client.responses.create(
        model = MODEL,
        input = (
        "Text: 'I love this product!' -> Sentiment: Positive\n"
        "Text: 'It broke after one day.' -> Sentiment: Negative\n"
        "Text: 'The delivery was three days late.' -> Sentiment:? "
    )
)
print("\n GPT: ", response.output_text)
```

```python
response = client.responses.create(
        model = MODEL,
        input = (" give me tips for a good presenstation")
)
print("\n GPT: ", response.output_text)
```

```python
response = client.responses.create(
        model = MODEL,
        input = (
        "List exactly 3 tips for a 5-minute student presentation. "
        "Format as a numbered list. Each tip must be one sentence, no explanations."
    )
)
print("\n GPT: ", response.output_text)
```

# Reasoning & Verbosity

```python
response = client.responses.create(
        model = MODEL,
        input = " what is 24*7",
        reasoning= {"effort":"minimal"},
        text={"verbosity":"low"}
)
print("\n GPT: ", response.output_text)
```

```python
response = client.responses.create(
        model = MODEL,
        input = " what is 24*7, show the reasoning",
        reasoning= {"effort":"high"},
        text={"verbosity":"high"}
)
print("\n GPT: ", response.output_text)
```
